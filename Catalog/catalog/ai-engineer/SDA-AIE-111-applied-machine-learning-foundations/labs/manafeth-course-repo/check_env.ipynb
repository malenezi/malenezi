```python
import sys, time
ok = True
def check(name, fn, hint):
    global ok
    try:
        detail = fn()
        print(f"✓ {name}" + (f" — {detail}" if detail else ""))
    except Exception as e:
        ok = False
        print(f"✗ {name}: {e}\n   fix: {hint}")

check("Python ≥ 3.10", lambda: (sys.version.split()[0],
      1/0 if sys.version_info < (3,10) else None) and sys.version.split()[0],
      "install Python 3.12 (64-bit)")
check("pandas", lambda: __import__("pandas").__version__, "pip install -r requirements.txt")
check("numpy", lambda: __import__("numpy").__version__, "pip install -r requirements.txt")
check("scikit-learn ≥ 1.5", lambda: __import__("sklearn").__version__, "pip install -r requirements.txt")
check("xgboost ≥ 2.0", lambda: __import__("xgboost").__version__,
      "pip install xgboost (Windows: try conda install -c conda-forge xgboost)")
check("matplotlib", lambda: __import__("matplotlib").__version__, "pip install -r requirements.txt")
check("pyarrow (parquet)", lambda: __import__("pyarrow").__version__, "pip install pyarrow")
check("openpyxl (capstone xlsx)", lambda: __import__("openpyxl").__version__, "pip install openpyxl")
```

```python
# Data reachable?
from pathlib import Path
import pandas as pd
def find_data():
    for cand in [Path.cwd(), *Path.cwd().parents]:
        p = cand / "data" / "manafeth_customers.parquet"
        if p.exists(): return p
    raise FileNotFoundError("copy the Manafeth data package into <repo>/data/")
check("manafeth_customers.parquet opens (48,000 × 19)", lambda: str(pd.read_parquet(find_data()).shape),
      "pull the course datasets from the shared drive into data/")
```

```python
# 10-second XGBoost fit
import numpy as np, time
from xgboost import XGBClassifier
def xgb_fit():
    rng = np.random.default_rng(0)
    X = rng.normal(size=(5000, 10)); y = (X[:, 0] + rng.normal(size=5000) > 0).astype(int)
    t0 = time.time()
    XGBClassifier(n_estimators=50, n_jobs=-1).fit(X, y)
    return f"{time.time()-t0:.1f}s"
check("XGBoost fits", xgb_fit, "reinstall xgboost; fallback: HistGradientBoostingClassifier for Day 3")
print("\nALL CHECKS PASSED — see you on Day 1." if ok else "\nSome checks failed — fix before Day 1.")
```
