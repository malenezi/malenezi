# Saudi SMS reminder RCT — worked solution
All calculations use the published aggregate 2×2 table.

```python
import math

sms_no_show, sms_n = 66, 251
control_no_show, control_n = 100, 251
p_sms = sms_no_show / sms_n
p_control = control_no_show / control_n
risk_difference = p_sms - p_control
risk_ratio = p_sms / p_control
odds_ratio = (sms_no_show / (sms_n - sms_no_show)) / (control_no_show / (control_n - control_no_show))
number_needed_to_remind = 1 / abs(risk_difference)
risk_difference, risk_ratio, odds_ratio, number_needed_to_remind
```

```python
# Two-sided pooled two-proportion z test
pooled = (sms_no_show + control_no_show) / (sms_n + control_n)
se0 = math.sqrt(pooled * (1 - pooled) * (1/sms_n + 1/control_n))
z = risk_difference / se0
p_two_sided = math.erfc(abs(z) / math.sqrt(2))
z, p_two_sided
```

```python
# Newcombe interval: combine the two Wilson score intervals
def wilson(x, n, zcrit=1.96):
    p = x / n
    d = 1 + zcrit**2 / n
    centre = p + zcrit**2 / (2*n)
    radius = zcrit * math.sqrt(p*(1-p)/n + zcrit**2/(4*n*n))
    return (centre-radius)/d, (centre+radius)/d

l_sms, u_sms = wilson(sms_no_show, sms_n)
l_control, u_control = wilson(control_no_show, control_n)
lower = risk_difference - math.sqrt((p_sms-l_sms)**2 + (u_control-p_control)**2)
upper = risk_difference + math.sqrt((u_sms-p_sms)**2 + (p_control-l_control)**2)
lower, upper
```

## Interpretation
Assignment to SMS reduced non-attendance by about 13.5 percentage points (95% CI approximately 5.3 to 21.5 points lower; two-sided p≈0.0013) among eligible patients in this clinic. The single-clinic, internal-medicine, Arabic-message, 2011 setting limits direct generalization to every Saudi clinic.
